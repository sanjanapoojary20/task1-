```python
import numpy as np
import pandas as pd
data = pd.read_csv("personalized_medication_dataset.csv")
print(data.head())
```

```python
# Fill missing categorical values with mode

data['Chronic_Conditions'].fillna(data['Chronic_Conditions'].mode()[0], inplace=True)

data['Drug_Allergies'].fillna(data['Drug_Allergies'].mode()[0], inplace=True)

data['Genetic_Disorders'].fillna(data['Genetic_Disorders'].mode()[0], inplace=True)

data['Recommended_Medication'].fillna(data['Recommended_Medication'].mode()[0], inplace=True)

data['Dosage'].fillna(data['Dosage'].mode()[0], inplace=True)

data['Duration'].fillna(data['Duration'].mode()[0], inplace=True)
```
